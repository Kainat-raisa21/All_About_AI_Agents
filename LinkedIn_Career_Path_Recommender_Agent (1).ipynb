```python
import google.generativeai as genai

model = genai.GenerativeModel(model_name='gemini-1.5-flash',
                                         system_instruction=[
        "You are a career advisor specializing in personalized career path recommendations.",
        "Your mission is to analyze LinkedIn profile data and user queries to suggest actionable career paths.",
    ],
        )
```

```python
def career_path_recommender(profile, query):
  prompt = f"""
  Your role: You are a career advisor specializing in personalized career path recommendations.
        Your mission is to analyze LinkedIn profile data and user queries to suggest actionable career paths.
    LinkedIn Profile Data:
    Name: {profile['Name']}
    Headline: {profile['Headline']}
    Location: {profile['Location']}
    About: {profile['About']}
    Top Skills: {profile['Top Skills']}
    Recent Post: {profile['Post']}

    User Query: {query}

    Provide a personalized career path recommendation for this individual.
    """

  response = model.generate_content([prompt])
  return response.text
```

```python
def linkedin_agent(dataframe, user_query):
  print(f"Analyzing profile: {dataframe['Name']}")
  recommendation = career_path_recommender(dataframe, user_query)
  print(f"Recommendation for {dataframe['Name']}:\n{recommendation}\n")
```

```python
# Example usage
if __name__ == "__main__":
    # Load the CSV file as a pandas dataframe
    csv_file_path = "linkedin_profile_posts_Debjoty Mitra.csv"
    df = pd.read_csv(csv_file_path)

    # User query
    user_query = input("Enter your query here:")

    # Run the LinkedIn agent
    linkedin_agent(df, user_query)
```
